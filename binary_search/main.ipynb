## Task 1 (easy): Search insert position

Given a sorted array of distinct integers and a target, return the index where the target is found. If not present, return the index where it would be inserted to keep the array sorted.

Must run in `O(log n)`.

Examples:
- `nums = [1, 3, 5, 6], target = 5` → `2`
- `nums = [1, 3, 5, 6], target = 2` → `1`
- `nums = [1, 3, 5, 6], target = 7` → `4`
- `nums = [1, 3, 5, 6], target = 0` → `0`

```python
import time

def search_insert(nums, target):

    h = len(nums)-1
    l = 0

    while l<=h:
        mid = (h+l)//2
        if nums[mid] == target:
            print("Found target: ", mid)
            return mid
        elif nums[mid] < target:
            l = mid + 1
        elif nums[mid] > target:
            h = mid - 1

    print('Nothing found')
    return None


search_insert([1, 3, 5, 6], 5)
search_insert([1, 3, 5, 6], 2)
search_insert([1, 3, 5, 6], 7)
search_insert([1, 3, 5, 6], 0)
print("Task 1 passed")
```
